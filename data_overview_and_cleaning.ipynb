# Data Overview and Cleaning

```python
# Import necessary libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
# Set display options for pandas
```

```python
# Load the dataset
df = pd.read_csv('datasets/womens-educational-attainment-vs-fertility/womens-educational-attainment-vs-fertility.csv')
# Display the first few rows of the dataset
df.head()
```

```python
# Display basic information about the dataset
df.info()
# Display summary statistics of the dataset
df.describe(include='all')
#display dataset dimensions
print("Dataset Dimensions:", df.shape)
```

```python
# Extract Column names
column_names = df.columns.tolist()
# Look for missing values
missing_values = df.isnull().sum()
# Display the column names and missing values
print("Column Names:", column_names)
print("Missing Values:\n", missing_values)
```

```python
# Check data types of each column
data_types = df.dtypes
# Display data types
print("Data Types:\n", data_types)
```

```python

# Check for duplicate rows
duplicate_rows = df.duplicated().sum()
# Display the number of duplicate rows
print("Number of Duplicate Rows:", duplicate_rows)
```

```python
# Visualize the distribution of each numeric column
numeric_columns = df.select_dtypes(include=[np.number]).columns.tolist()
for column in numeric_columns:
    plt.figure(figsize=(10, 5))
    sns.histplot(df[column], kde=True)
    plt.title(f'Distribution of {column}')
    plt.xlabel(column)
    plt.ylabel('Frequency')
    plt.show()
```
